## Set-up the project environment for Formula1 Project
1. Create External Location 'db_formula1_ext_dl_formula1-incr'
2. Create catalog formula1_incr
3. Create Schemas landing, bronze, silver and gold
4. Create Volume Files in the landing schema

#### Access Cloud Storage

```sql
%fs ls 'abfss://formula1-incr@dbformula1extdl.dfs.core.windows.net/landing/'
```

#### Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS db_formula1_ext_dl_formula1_incr
URL 'abfss://formula1-incr@dbformula1extdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-formula1-sc`)
COMMENT 'External location for formula1-incr container'
```

#### Create Catalog formula1_incr

```sql
SHOW CATALOGS;
```

#### Create Schemas landing, bronze, silver and gold

```sql
CREATE CATALOG IF NOT EXISTS formula1_incr
MANAGED LOCATION 'abfss://formula1@dbformula1extdl.dfs.core.windows.net/'
COMMENT 'This is the main catalog fro formula1 project';
```

```sql
CREATE DATABASE IF NOT EXISTS formula1_incr.landing;

CREATE DATABASE IF NOT EXISTS formula1_incr.bronze
MANAGED LOCATION 'abfss://formula1-incr@dbformula1extdl.dfs.core.windows.net/bronze';

CREATE DATABASE IF NOT EXISTS formula1_incr.silver
MANAGED LOCATION 'abfss://formula1-incr@dbformula1extdl.dfs.core.windows.net/silver';

CREATE DATABASE IF NOT EXISTS formula1_incr.gold
MANAGED LOCATION 'abfss://formula1-incr@dbformula1extdl.dfs.core.windows.net/gold';
```

#### Create Volume Files

```sql
CREATE EXTERNAL VOLUME formula1_incr.landing.files
LOCATION 'abfss://formula1-incr@dbformula1extdl.dfs.core.windows.net/landing'
COMMENT 'landing files'
```
